```sql
with team_metrics as (

select 
    constructor_name,
    sum(races) as race_starts,
    sum(number_of_wins) as wins,
    sum(number_of_podiums) as podiums,
    sum(case when ranks = 1 then 1 else 0 end) as championships
from formula1.gold.constructor_stats
group by constructor_name
having wins >=1
)
select 
    constructor_name,
    race_starts,
    wins,
    podiums,
    championships,
    (championships * 100) + (wins * 10) + (podiums * 1) as score 
from team_metrics
order by score desc
limit 20
```
