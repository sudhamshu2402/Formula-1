```sql
with driver_metrics as (

select 
    driver_name,
    sum(races_starts) as race_starts,
    sum(number_of_wins) as wins,
    sum(number_of_podiums) as podiums,
    sum(case when standing = 1 then 1 else 0 end) as championships
from formula1.gold.v_driver
group by driver_name
having wins >=1
)
select 
    driver_name,
    race_starts,
    wins,
    podiums,
    championships,
    (championships * 100) + (wins * 10) + (podiums * 3) as score
from driver_metrics
order by score desc
limit 20
```
